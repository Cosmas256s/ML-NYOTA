## Final Model Export & On-Device Inference Function

**Goal**: Produce the exact JSON your Flutter app can load and a Dart-ready inference equivalent.

**Focus**:
- Lightweight deployment artifact (no heavy runtime ML package).
- Fast per-interaction update for on-device personalization.
- Autism-aware adaptation using forgetting, hints, and behavior.

```python
# CELL 1: Final Export to JSON
# JSON module for writing deployment artifact
import json
# Path for robust file path handling
from pathlib import Path

# Build final, app-facing parameter package
final_model = {
    "skills": ["counting", "shape_recognition", "addition", "subtraction", "multiplication", "division"],
    "parameters": {
        "P_L0": 0.451,
        "P_T": 0.142,
        "P_G": 0.25,
        "P_S": 0.22,
        "P_F": 0.112,        # forgetting
        "alpha_h": 0.6,      # hints
        "beta_b": 1.2        # behavior
    }
}

# Ensure model directory exists
MODELS_DIR = Path("../models")
MODELS_DIR.mkdir(parents=True, exist_ok=True)

# Save final model JSON for Flutter assets
with open(MODELS_DIR / "final_model.json", "w") as f:
    json.dump(final_model, f, indent=2)

# Confirmation message
print("Final model exported – ready for Flutter!")
```

```python
# CELL 2: Exact Inference Function (Copy-Paste Logic to Dart)
def bkt_inference(prior, correct, hint_used, behavior_score, params):
    """Exact update rule to mirror in Flutter (Dart)."""
    # Unpack model parameters for readability
    pG = params["P_G"]
    pS = params["P_S"]
    pT = params["P_T"]
    pF = params["P_F"]
    alpha_h = params["alpha_h"]
    beta_b = params["beta_b"]

    # Adjust learning transition using behavior and hint support
    effective_pT = pT * (beta_b * behavior_score) * (1 + alpha_h * hint_used)
    # Adjust forgetting inversely with behavior
    effective_pF = pF * (1 - beta_b * behavior_score)

    # Bayesian update after observing correctness
    if correct:
        posterior = (prior * (1 - pS)) / (prior * (1 - pS) + (1 - prior) * pG)
    else:
        posterior = (prior * pS) / (prior * pS + (1 - prior) * (1 - pG))

    # Propagate to next-step prior with learning and forgetting
    new_prior = posterior * (1 - effective_pF) + (1 - posterior) * effective_pT

    # Return rounded probability for clean app logging/debugging
    return round(new_prior, 4)

# Quick test call to validate function behavior
test_prior = 0.65
result = bkt_inference(
    test_prior,
    correct=True,
    hint_used=1,
    behavior_score=0.8,
    params=final_model["parameters"]
)
print("New knowledge probability:", result)
```
